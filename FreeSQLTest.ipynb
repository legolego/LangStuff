```python
import oracledb
import os
from dotenv import load_dotenv

load_dotenv()  # loads variables from a .env file in this directory, if present

user = os.environ.get("FREE_SQL_USER")
password = os.environ.get("FREE_SQL_PASSWORD")
local_dsn = os.environ.get("FREE_SQL_DSN")

connection = oracledb.connect(
    user=user,
    password=password,
    dsn=local_dsn)

print("Successfully connected to Oracle Database")

cursor = connection.cursor()
for result in cursor.execute("SELECT * FROM OLYM.OLYM_ATHLETES where rownum < 10"):
    print(result)
```
